# Pythagoras pie puzzle
---------------------------------------------
**Problem Statement**

*A gigantic pie is split amongst 100 guests. The first guest gets 1% of the pie. Then the second guest gets 2% of the remaining pie. Then the third guest gets 3% of the remaining pie, and so on, until the last guest gets 100% of the remaining pie.*

Which guest gets the largest piece of pie?

Usually, the puzzle focuses on identifying the guest who receives the largest piece of the pie. However, out of curiosity, I delved deeper—visualizing how the pie was divided and analyzing how many guests received the majority of it.

----------------------------------------------

**Analysis Made**

- 10th Guest will get maximum split of 6.28%
- 100th Guest will get minimum split of 9.33e-41 %
- First 24 guests(24% of Guests) will get atleast 1% split
- First 24 guests(24% of Guests) will get 96.24% of the pie
- Other guests(76% of Guests) all together will get only 3.76%
--------------------------------------------------


```python
pie = 100.00
guest_percent =[]
guest=[]
for i in range(1,101):
    percent = (pie/100) * i
    pie = pie - percent
    print("guest {0} recieve {1} percentage, remaining pie: {2}".format(i,percent,pie))
    guest_percent.append(percent)
    guest.append(str(i))
```

```python
print(guest_percent)
print(guest)
max_value = max(guest_percent)
max_index = guest_percent.index(max_value)

min_value = min(guest_percent)
min_index = guest_percent.index(min_value)

print("Max split", max_value, "Guest Number", max_index+1)
print("Min split", min_value, "Guest Number", min_index+1)
```

```python
import matplotlib.pyplot as plot
def graph():
    top_n = 24
    top_data = guest_percent[:top_n]
    other_data = sum(guest_percent[top_n:])
    
    pie_data = top_data + [other_data]
    labels = [f"Slice {i+1}" for i in range(top_n)] + ["Others"]

    plot.figure(figsize=(10, 10))
    b = plot.pie(
        pie_data,
        labels=labels,
        textprops={'fontsize': 14},
        autopct='%1.1f%%' 
    )
    plot.title("pythagoras pie puzzle", fontsize=20)
    plot.legend(b[0], labels, fontsize=12)
    plot.show()
graph()
```

```python
def graph_line():
    plot.figure(figsize=(10, 6))
    plot.plot(guest_percent, linestyle='dotted')
    plot.title('Guest Percent', fontsize=16)
    plot.xlabel('Guest', fontsize=12)
    plot.ylabel('share', fontsize=12)
    plot.grid(True, which="both", linestyle='--', linewidth=0.5)
    plot.show()
graph_line()
```
